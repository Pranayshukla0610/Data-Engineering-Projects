```python
#Remove Duplicates from a Sorted Array
#Problem

#Given a sorted array, remove duplicate elements in-place.

#Input:

#[1, 1, 2, 2, 3, 4, 4]

#Output:

#[1, 2, 3, 4]

arr = list(map(int, input("Enter an array:").split()))

result = []

for i in range(len(arr)):
  if arr[i] != arr[i-1]:
    result.append(arr[i])

print(result)
```

**OR**

```python
arr = list(map(int, input("Enter an array:").split()))

result = []

left = 0

for right in range(len(arr)):
  if arr[right] != arr[left]:
    left += 1
    arr[left] = arr[right]

result = arr[:left+1]

print("After removing duplicates:", result)
```

```python
#Two Sum
#Problem

#Given an array and a target value, find two numbers whose sum equals the target.

#Input:

#Array = [2, 7, 11, 15]
#Target = 9

#Output:

#[0, 1]

arr = list(map(int, input("Enter an array:").split()))
target = int(input("Enter a target value"))

result = []

for i in range(len(arr)):
  for j in range(i+1,len(arr)):
    if arr[i] + arr[j] == target:
      result.append([i,j])

print(result)
```

**OR**

```python
arr = list(map(int, input("Enter an array:").split()))
target = int(input("Enter a target value"))

seen = {}

for i in range(len(arr)):
  complement = target - arr[i]
  if complement in seen:
    print("Indices:",[seen[complement],i])
    print("Values:",[complement,arr[i]])
    break
  seen[arr[i]] = i
```

```python
#Maximum Subarray Sum — Kadane's Algorithm
#Input:

#[-2, 1, -3, 4, -1, 2, 1, -5, 4]

#The maximum subarray is:

#[4, -1, 2, 1]

#Sum:

#4 + (-1) + 2 + 1 = 6

arr = list(map(int, input("Enter an array:").split()))

max_sum = arr[0]
curr_sum = arr[0]

for i in range(1,len(arr)):
  curr_sum = max(arr[i],curr_sum+arr[i])
  max_sum = max(max_sum,curr_sum)

print("Maximum subarray sum:",max_sum)
```

```python
#Find Missing Number
#Input:

#[3, 0, 1]

#Numbers should be:

#0, 1, 2, 3

#Missing:

#2

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)

actual_sum = sum(arr)
expected_sum = n * (n + 1)//2

missing = expected_sum - actual_sum

print("Array:",arr)
print("Missing Number:",missing)
```

```python
#Best Time to Buy and Sell Stock
#Find the maximum profit you can make by buying once and selling once.

#Input:

#[7, 1, 5, 3, 6, 4]

#Buy at:

#1

#Sell at:

#6

#Profit:

#6 - 1 = 5

arr = list(map(int, input("Enter an array:").split()))

min_num = arr[0]
max_profit = 0

for i in range(1,len(arr)):
  min_num = min(min_num,arr[i])
  max_profit = max(max_profit,arr[i]-min_num)

print(max_profit)
```
