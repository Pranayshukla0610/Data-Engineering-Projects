```python
#Longest Increasing Subsequence (LIS)
#Input
#10 9 2 5 3 7 101 18
#Output
#4

#Subsequence:

#2 3 7 101

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)
dp = [1] * n

for i in range(n):
  for j in range(i):
    if arr[i] > arr[j]:
      dp[i] = max(dp[i], dp[j]+1)

print("Longest Increasing Subsequence:",max(dp))
```

```python
#Leaders in an Array
#An element is called a Leader if it is greater than or equal to all the elements to its right.

#The last element is always a leader.

#Example

#Input

#16 17 4 3 5 2

#Output

#Leaders:
#17 5 2

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)

leaders = []

max_right = arr[-1]
leaders.append(max_right)

for i in range(n-2, -1, -1):
  if arr[i] >= max_right:
    leaders.append(arr[i])
    max_right = arr[i]

leaders.reverse()
print("Leaders:",leaders)
```

```python
#Find Equilibrium Index ⭐⭐⭐
#Problem

#Find an index where

#Left Sum = Right Sum
#Example

#Input

#1 3 5 2 2

#Output

#Equilibrium Index = 2

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)
found = 0

left_sum = 0
right_sum = sum(arr)

for i in range(n):
  right_sum -= arr[i]
  if left_sum == right_sum:
    print("Equilibrium Index:",i)
    break

  left_sum += arr[i]

if not found:
  print("Equilibrium Index not found")
```

```python
#Maximum Difference (Larger After Smaller)
#Input
#2 3 10 6 4 8 1
#Output
#Maximum Difference = 8

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)
min_val = arr[0]
max_diff = arr[1] - arr[0]

for i in range(1,n):
  max_diff = max(max_diff, arr[i] - min_val)
  min_val = min(min_val, arr[i])

print("Maximum Difference:",max_diff)
```
