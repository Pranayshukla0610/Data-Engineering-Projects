```python
#Remove Duplicates from a Sorted Array

arr = list(map(int, input("Enter an array:").split()))
arr.sort()
result = [arr[0]]

for i in range(1,len(arr)):
  if arr[i] != arr[i-1]:
    result.append(arr[i])

print("Array after removing duplicates:",result)
```

```python
#Remove All Occurrences of an Element
arr = list(map(int, input("Enter an array:").split()))
target = int(input("Enter a target:"))

result = []

for i in range(len(arr)):
  if arr[i] != target:
    result.append(arr[i])

print("Result after removing all occurences:",result)
```

```python
#Find Missing Number (1 to n)

arr = list(map(int, input("Enter an array:").split()))
n = len(arr) + 1

expected_sum = n * (n + 1)//2

actual_sum = sum(arr)

missing_number = expected_sum - actual_sum

print("Missing Number:",missing_number)
```

```python
#Find All Pairs with Given Sum

#Input
#1 5 7 -1 5

#6
#Output
#Pairs:
#1 5
#7 -1
#1 5

arr = list(map(int, input("Enter an array:").split()))

target = int(input("Enter a target:"))

pairs = []

for i in range(len(arr)):
  for j in range(i+1,len(arr)):
    if arr[i] + arr[j] == target:
      pairs.append((arr[i],arr[j]))

print("All pairs:",pairs)
```

```python
#Find Maximum Difference (Larger Element After Smaller)

arr = list(map(int, input("Enter an array:").split()))

minimum = arr[0]
maximum_diff = arr[1] - arr[0]

for i in range(1, len(arr)):
  difference = arr[i] - minimum
  if difference > maximum_diff:
    maximum_diff = difference

  if arr[i] < minimum:
    minimum = arr[i]

print("Maximum Difference:",maximum_diff)
```

```python
#Equilibrium Index
#Index where left sum equals right sum.
#Input
#1 3 5 2 2
#Output
#Equilibrium Index = 2

arr = list(map(int, input("Enter array: ").split()))

left_sum = 0
right_sum = sum(arr)

for i in range(len(arr)):
  right_sum -= arr[i]
  if left_sum == right_sum:
    print("Equilibrium Index:")
    break

  left_sum += arr[i]

print(i)
```

```python
#Reverse an Array (In-place)
#Input
#1 2 3 4 5
#Output
#Reversed Array: [5, 4, 3, 2, 1]

arr = list(map(int, input("Enter an array:").split()))

l = 0
r = len(arr)-1

while l < r:
  arr[l], arr[r] = arr[r], arr[l]
  l += 1
  r -= 1

print("Reversed Array:",arr)
```
