```python
#Minimum Subarray Sum

arr = list(map(int, input("Enter an array:").split()))

current_sum = arr[0]
minimum_sum = arr[0]

for i in range(1,len(arr)):
  current_sum = min(arr[i],current_sum+arr[i])
  minimum_sum = min(current_sum, minimum_sum)

print("Minimum Subarray Sum:",minimum_sum)
```

```python
#Check Whether Two Arrays are Rotations

arr1 = list(map(int, input("Enter an array:").split()))
arr2 = list(map(int, input("Enter an array:").split()))

if len(arr1) != len(arr2):
  print("No Rotation")

else:
  doubled = arr1 + arr1
  found = False

  for i in range(len(arr1)):
    if doubled[i:i+len(arr1)] == arr2:
      found = True
      break

  if found:
    print("Rotation")
  else:
    print("No Rotation")
```

```python
#Find Maximum Consecutive Ones in this binary array

arr = list(map(int, input("Enter an array:").split()))

count = 0
max_count = 0

for num in arr:
  if num == 1:
    count += 1
    max_count = max(count,max_count)
  else:
    count = 0

print("Maximum Consecutive Ones:",max_count)
```

```python
#Given an integer array, return an array where each element is the product of all other elements except itself

#Input
#1 2 3 4
#Output
#[24, 12, 8, 6]

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)
l = [1]*n
r = [1]*n
result = [1]*n

for i in range(1,n):
  l[i] = l[i-1] * arr[i-1]

for i in range(n-2,-1,-1):
  r[i] = r[i+1] * arr[i+1]

for i in range(n):
  result[i] = l[i] * r[i]

print("Product Array:",result)
```

```python
#Container With Most Water
#Input
#1 8 6 2 5 4 8 3 7
#Output
#49

arr = list(map(int, input("Enter an array:").split()))

l = 0
r = len(arr) - 1
max_area = 0

while l < r:
  width = r - l
  height = min(arr[l],arr[r])
  area = width * height

  max_area = max(area, max_area)

  if arr[l] < arr[r]:
    l += 1
  else:
    r -= 1

print("Maximum Water:",max_area)
```

```python
#Range Sum Query – Immutable
#Input

#Array:
#-2 0 3 -5 2 -1

#Query:
#0 2

#Output

#1


arr = list(map(int, input("Enter array elements: ").split()))

n = len(arr)

# Create Prefix Sum Array
prefix = [0] * n

prefix[0] = arr[0]

for i in range(1, n):
    prefix[i] = prefix[i - 1] + arr[i]

print("\nPrefix Sum Array:", prefix)


q = int(input("\nEnter number of queries: "))

for _ in range(q):

    left, right = map(int, input("Enter Left and Right Index: ").split())

    if left == 0:
        answer = prefix[right]
    else:
        answer = prefix[right] - prefix[left - 1]

    print(f"Sum from index {left} to {right} = {answer}")
```
