```python
#Calculate the Sum from 1 to N
#Problem: Given an integer n, calculate the sum of all numbers from 1 to n using DP.
#Input
#n = 5


#Expected output
#15

def calculate_sum(n):
  if n < 1:
    return 0

  dp = [0] * (n + 1)
  dp[0]= 0

  for i in range(1,n+1):
    dp[i] = dp[i - 1] + i

  return dp[n]

n = 5
print(calculate_sum(n))
```

```python
#Calculate Factorial Using DP
#Problem: Given an integer n, calculate its factorial using a DP list.

#Input
#n = 5


#Expected output
#120

def calculate_fact(n):
  if n < 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = 0
  dp[1] = 1

  for i in range(2, n+1):
    dp[i] = i * dp[i - 1]

  return dp[n]

n = 5
print(calculate_fact(n))
```

```python
#Find the N-th Number in a Sequence Starting at 2
#Problem: The sequence is 2, 4, 6, 8, 10, .... Given n, find the n-th number using DP.
#Input
#n = 5


#Expected output
#10

def even_sequence(n):
  if n <= 0:
    return 0

  dp = [0] * (n + 1)
  dp[1] = 2

  for i in range(2, n+1):
    dp[i] = dp[i - 1] + 2

  return dp[n]

n = 5
print(even_sequence(n))
```

```python
#Count Ways to Reach Step N Using Only One-Step Jumps
#Problem: You are standing at step 0. You can move forward exactly one step at a time. Find the number of ways to reach step n.
#Input
#n = 5


#Expected output
#1

def count_steps(n):
  if n <= 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = 1
  dp[1] = 1

  for i in range(2, n + 1):
    dp[i] = dp[i - 1]

  return dp[n]

n = 5
print(count_steps(n))
```

```python
#Count Ways to Reach N Using Jumps of 1, 2, or 3
#Problem: Starting at step 0, you can jump forward by 1, 2, or 3 steps. Find the number of ways to reach step n.
#Input
#n = 3


#Expected output
#4

def count_jump(n):
  if n < 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(1, n+1):
    for jump in [1,2,3]:
      if i >= jump:
        dp[i] += dp[i - jump]

  return dp[n]

n = 3
print(count_jump(n))
```
