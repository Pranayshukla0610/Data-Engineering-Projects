```python
#Fibonacci Number Using Tabulation
#Level: Beginner
#Problem: Given an integer n, calculate the \(n\)-th Fibonacci number using recursion and memoization. The first two numbers are 0 and 1.
#Input
#n = 6


#Output
#8

def fibonacci(n):
  if n == 0:
    return 0

  dp = [0] *(n + 1)
  dp[0] = 0
  dp[1] = 1

  for i in range(2, n+1):
    dp[i] = dp[i - 1] + dp[i - 2]

  return dp[n]

n = 6
print(fibonacci(n))
```

```python
#Fibonacci Number Using Memoization
#Level: Beginner
#Problem: Given an integer n, calculate the \(n\)-th Fibonacci number using recursion and memoization. The first two numbers are 0 and 1.
#Input
#n = 6


#Output
#8

def fibonacci(n, memo=None):
  if memo is None:
    memo = {}

  if n <= 1:
    return n

  if n in memo:
    return memo[n]

  memo[n] = fibonacci(n - 1, memo) + fibonacci(n - 2, memo)

  return memo[n]

n = 6
print(fibonacci(n))
```

```python
#Climbing Stairs with 1, 2, or 3 Steps
#Level: Beginner
#Problem: You can climb either 1, 2, or 3 steps at a time. Find the number of distinct ways to reach the top of a staircase containing n steps.
#Input
#n = 4


#Output
#7

def climbing_stairs(n):
  if n == 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(1, n + 1):
    for jump in (1,2,3):
      if i >= jump:
        dp[i] += dp[i - jump]

  return dp[n]

n = 4
print(climbing_stairs(n))
```

```python
#Minimum Cost to Reach the Last Position
#Level: Beginner
#Problem: Given an array where each element represents the cost of visiting a position, start at index 0 and reach the last index. You may move either one or two positions forward. Find the minimum total cost, including the starting and ending positions.
#Input
#cost = [1, 3, 2, 4, 6]


#Output
#9

def min_cost(cost):
  n = len(cost)

  if n == 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = cost[0]

  if n > 1:
    dp[1] = cost[0] + cost[1]

  for i in range(2,n):
    dp[i] = cost[i] + min(dp[i - 1], dp[i - 2])

  return dp[n - 1]

cost = [1, 3, 2, 4, 6]
print(min_cost(cost))
```

```python
#House Robber — Maximum Money
#Level: Beginner
#Problem: A robber is given an array representing money in each house. They cannot rob two adjacent houses. Find the maximum amount of money they can collect.
#Input
#nums = [2, 7, 9, 3, 1]


#Output
#12

def max_amt(nums):
  n = len(nums)
  if n == 0:
    return 0

  dp = [0] * n
  dp[0] = nums[0]

  for i in range(1, n):
    skip = dp[i - 1]

    rob = nums[i]

    if i >= 2:
      rob += dp[i - 2]

    dp[i] = max(skip,rob)

  return dp[-1]

nums = [2, 7, 9, 3, 1]
print(max_amt(nums))
```
