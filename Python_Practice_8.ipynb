```python
#Find the Intersection of Two Lists (Without Using Sets)

list1 = list(map(int,input("Enter first list:").split()))
list2 = list(map(int, input("Enter second list:").split()))

result = []

for num in list1:
  if num in list2:
    result.append(num)

print(result)
```

```python
#Count Words in a Sentence

sentence = input("Enter a sentence:")

words = sentence.split()
count = 0

for word in words:
  count += 1

print(count)
```

```python
sentence = input("Enter a sentence:")

words = sentence.split()

count = {}

for word in words:
    count[word] = count.get(word, 0) + 1

print(count)
```

```python
#Check if a Number is a Perfect Number

number = int(input("Enter a number:"))

total = 0

for i in range(1,number):
  if number % i == 0:
    total += i

if total == number:
  print("Perfect number")
else:
  print("Not a Perfect Number")
```

```python
#Find the Largest Word in a Paragraph
text = input("Enter a sentence:")

words = text.split()
maximum = ""

for word in words:
  if len(word) > len(maximum):
    maximum = word

print("Largest Word:",maximum)
```

```python
#Find difference between two lists

list1 = list(map(int,input("Enter first list:").split()))
list2 = list(map(int, input("Enter second list:").split()))

result = []

for num in list1:
  if num not in list2:
    result.append(num)

print(result)
```

```python
#Find All Leaders in an Array

#A leader is an element greater than all elements to its right.

arr = list(map(int, input("Enter an array").split()))

leaders = []

maximum = float('-inf')

for num in reversed(arr):
  if num > maximum:
    maximum = num
    leaders.append(num)

leaders.reverse()
```
