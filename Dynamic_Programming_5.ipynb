```python
#Calculate the Sum of the First N Odd Numbers
#Problem: Given n, calculate the sum of the first n odd numbers using DP.
#Input
#n = 5


#Expected output
#25

def sum_odd(n):
  if n == 0:
    return 0

  dp = [0] * (n + 1)


  for i in range(1, n+ 1):
    dp[i] = dp[i - 1] + 2 * i - 1

  return dp[n]

n = 5
print(sum_odd(n))
```

```python
#Calculate the Sum of the First N Even Numbers
#Problem: Calculate the sum of the first n positive even numbers using DP.
#Input
#n = 4


#Expected output
#20

def sum_even(n):
  if n == 0:
    return 0

  dp = [0] * (n + 1)

  for i in range(1, n + 1):
    dp[i] = dp[i - 1] + 2 * i

  return dp[n]

n = 4
print(sum_even(n))
```

```python
#Calculate Powers of 2
#Problem: Given n, calculate \(2^n\) using DP.
#Input
#n = 5


#Expected output
#32

def power_of_two(n):
  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(1, n + 1):
    dp[i] = dp[i - 1] * 2

  return dp[n]

n = 5
print(power_of_two(n))
```

```python
#Calculate Powers of 3
#Problem: Given n, calculate \(3^n\) using DP.
#Input
#n = 4


#Expected output
#81

def power_of_three(n):
  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(1, n + 1):
    dp[i] = dp[i - 1] * 3

  return dp[n]

n = 4
print(power_of_three(n))
```

```python
#Calculate the Product of Numbers from 1 to N
#Problem: Given n, calculate the product \(1 \times 2 \times \cdots \times n\) using DP. This is also called factorial.
#Input
#n = 4


#Expected output
#24

def product_of_numbers(n):
  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(1, n+1):
    dp[i] = dp[i - 1] * i

  return dp[n]

n = 4
print(product_of_numbers(n))
```
