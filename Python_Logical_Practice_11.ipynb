```python
#Find Leaders in an Array

#A leader is an element greater than all elements to its right.

#Input:
#16 17 4 3 5 2

#Output:
#17 5 2

arr = list(map(int, input("Enter an array:").split()))
leaders = []
max_right = arr[-1]
leaders.append(max_right)

for i in range(len(arr)-2, -1, -1):
  if arr[i] > max_right:
    max_right = arr[i]
    leaders.append(arr[i])

leaders.reverse()
print("Leaders:",leaders)
```

```python
#Find Equilibrium Index

#Find an index where:

#Left Sum = Right Sum

arr = list(map(int, input("Enter an array:").split()))

total_sum = sum(arr)
left_sum = 0

for i in range(len(arr)):
  total_sum -= arr[i]
  if left_sum == total_sum:
    print("Equilibrium Index:",i)
    break
  left_sum += arr[i]
else:
  print("No Equilibrium Index")
```

```python
#Find Maximum Consecutive Ones
#Input:
#1 1 0 1 1 1

#Output:
#3

arr = list(map(int, input("Enter an array:").split()))

max_count = 0
count = 0

for i in range(len(arr)):
  if arr[i] == 1:
    count += 1
  else:
    count = 0

  max_count = max(max_count, count)

print("Maximum Consecutive Ones:",max_count)
```

```python
#Find the First Missing Positive Integer
#Input

#3 4 -1 1

#Output

#2

arr = list(map(int, input("Enter an array:").split()))

positive_int = set()

for num in arr:
  if num > 0:
    positive_int.add(num)

missing = 1

while missing in positive_int:
  missing += 1

print("First Missing Positive Integer:",missing)
```

```python
#Find Maximum Subarray Sum (Kadane's Algorithm)
#Input

#-2 1 -3 4 -1 2 1 -5 4

#Output

#6

arr = list(map(int, input("Enter an array:").split()))

curr_sum = arr[0]
max_sum = arr[0]

for i in range(1, len(arr)):
  curr_sum = max(arr[i], curr_sum + arr[i])
  max_sum = max(max_sum, curr_sum)

print("Maximum Subarray Sum:",max_sum)
```
