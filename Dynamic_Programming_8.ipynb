```python
#Maximum Sum of a Subarray
#Easy · Kadane's Algorithm / DP
#Problem statement: Given an array containing positive and negative integers, find the maximum possible sum of a non-empty contiguous subarray.

#Input
#arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]


#Output
#6

def max_sum(arr):
  n = len(arr)
  if n == 0:
    return 0

  dp = [0] * n
  dp[0] = arr[0]
  for i in range(1,n):
    dp[i] = max(dp[i - 1] + arr[i], arr[i])
  return max(dp)

arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(max_sum(arr))
```

```python
#Climbing Stairs
#Easy · 1D DP
#Problem statement: You are climbing a staircase with n steps. At each move, you can climb either one step or two steps. Find the number of distinct ways to reach the top.

#Input
#n = 4


#Output
#5

def climbing_stairs(n):
  if n == 0:
    return 0

  if n <= 2:
    return n

  dp = [0] * (n + 1)
  dp[1] = 1
  dp[2] = 2

  for i in range(3,n+1):
    dp[i] = dp[i - 1] + dp[i - 2]
  return dp[n]

n = 4
print(climbing_stairs(n))
```

```python
#Sum of an Array Using DP
#Easy · Basic DP
#Problem statement: Given an array of integers, calculate the sum of all elements using a DP array that stores the cumulative sum up to each position.

#Input
#arr = [2, 4, 6, 8]


#Output
#20

def sum_arr(arr):
  n = len(arr)
  if n == 0:
    return 0

  dp = [0] * n
  dp[0] = arr[0]

  for i in range(1,n):
    dp[i] = dp[i - 1] + arr[i]

  return dp[n - 1]

arr = [2, 4, 6, 8]
print(sum_arr(arr))
```

```python
#Minimum Cost to Reach the End of an Array
#Easy · 1D DP
#Problem statement: You are given an array of non-negative costs. Start at index 0. At each move, you can jump one or two positions forward. Find the minimum total cost required to reach the
#last index.

#The cost of the starting and ending positions is included.

#Input
#cost = [1, 3, 2, 4]


#Output
#3


#Explanation: Jump from index 0 to index 2, then to index 3. The total cost is 1 + 2 + 4 = 7, but that is not optimal. Instead, jump from index 0 to index 1, then to index 3,
#costing 1 + 3 + 4 = 8. Let's calculate the true minimum through DP.

def min_cost(cost):
  n = len(cost)

  if n == 0:
    return 0

  if n == 1:
    return dp[0]


  dp = [0] * n
  dp[0] = cost[0]
  dp[1] = cost[1] + dp[0]

  for i in range(1,n):
    dp[i] = min(dp[i - 1], dp[i - 2]) + cost[i]

  return dp[n - 1]

cost = [1, 3, 2, 4]
print(min_cost(cost))
```

```python
#Number of Ways to Reach a Target Using 1 or 2
#Easy · Basic Recurrence
#Problem statement: Starting from zero, count the number of sequences of moves that add up to a target n. Every move can add either 1 or 2.

#Input
#n = 3


#Output
#3

def target_ways(n):
    if n < 0:
        return 0

    dp = [0] * (n + 1)
    dp[0] = 1

    for i in range(1, n + 1):
        dp[i] = dp[i - 1]

        if i >= 2:
            dp[i] += dp[i - 2]

    return dp[n]


n = 3
print(target_ways(n))
```

```python
#Maximum Sum of Non-Adjacent Elements
#Easy · 1D DP
#Problem statement: Given an array of non-negative integers, find the maximum sum you can obtain without selecting two adjacent elements.

#Input
#arr = [2, 7, 9, 3, 1]


#Output
#12

def max_non_adjacent_sum(arr):
    n = len(arr)

    if n == 0:
        return 0

    if n == 1:
        return arr[0]

    dp = [0] * n
    dp[0] = arr[0]
    dp[1] = max(arr[0], arr[1])

    for i in range(2, n):
        take = arr[i] + dp[i - 2]
        skip = dp[i - 1]

        dp[i] = max(take, skip)

    return dp[n - 1]


arr = [2, 7, 9, 3, 1]
print(max_non_adjacent_sum(arr))
```
