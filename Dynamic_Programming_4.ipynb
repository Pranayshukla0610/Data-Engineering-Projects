```python
#Calculate the Cumulative Product of an Array
#Problem: Given an array of numbers, create a DP list where each position stores the product of all elements from the beginning up to that position.
#Input
#arr = [2, 3, 4, 5]


#Expected output
#[2, 6, 24, 120]

def cumulative_product(arr):
  if len(arr) == 0:
    return 0

  n = len(arr)


  dp = [0] * (n)
  dp[0] = arr[0]

  for i in range(1, n):
    dp[i] = dp[i - 1] * arr[i]

  return dp

print(cumulative_product([2, 3, 4, 5]))
```

```python
#Count the Number of Digits in a Number Using DP
#Problem: Given a positive integer, create a DP list where each position stores the number of digits in the number formed by the first i digits of its string representation.
#Input
#num = 5832


#Expected output
#4

def count_digits(num):
  if num == 0:
    return 0

  n = len(str(num))

  dp = [0] * (n)
  dp[0] = 1
  for i in range(1,n):
    dp[i] = dp[i - 1] + 1

  return dp[n-1]

num = 5832
print(count_digits(num))
```

```python
#Find the Previous Number Minus 1
#Problem: A sequence starts with 10. Each new number is one less than the previous number. Return the n-th number, counting the first number as position 1.
#Input
#n = 5


#Expected output
#6

def previous_number(n):
  if n == 0:
    return 0
  dp = [0] * (n+1)
  dp[1] = 10

  for i in range(2,n + 1):
    dp[i] = dp[i - 1] - 1

  return dp[n]

n = 5
print(previous_number(n))
```

```python
#Find the Previous Number Plus 5
#Problem: Start a sequence with 5. Each new number is the previous number plus 5. Return the n-th number, counting the first number as position 1.
#Input
#n = 5


#Expected output
#25

def previous_number(n):
  if n == 0:
    return 0

  dp = [0] * (n + 1)
  dp[1] = 5

  for i in range(2, n + 1):
    dp[i] = dp[i - 1] + 5

  return dp[n]

n = 5
print(previous_number(n))
```

```python
#Count Ways to Reach N Using Only 2-Step Jumps
#Problem: Start at position 0. You can jump exactly 2 positions each time. Find the number of ways to reach position n.
#Input
#n = 6


#Expected output
#1

def two_step_jump(n):
  if n == 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(2, n + 1,2):
    dp[i] = dp[i - 2]

  return dp[n]

n = 6
print(two_step_jump(n))
```
