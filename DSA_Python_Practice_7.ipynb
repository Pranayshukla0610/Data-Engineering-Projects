```python
#Longest Mountain in an Array (LeetCode 845)

#Input
#2 1 4 7 3 2 5
#Output
#5

#Mountain

#1 4 7 3 2

#Length

#5

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)
maximum = 0

for i in range(1, n-1):
  if arr[i] > arr[i-1] and arr[i] > arr[i+1]:
    left = i
    right = i

    while left > 0 and arr[left] > arr[left-1]:
      left -= 1
    while right < n-1 and arr[right] > arr[right+1]:
      right += 1

    length = right - left + 1
    maximum = max(maximum, length)

print("Longest Mountain Array:",maximum)
```

```python
#Find the Longest Increasing Contiguous Subarray

#Input

#1 2 3 2 4 5 6 1

#Output

#4

#Subarray:

#2 4 5 6

arr = list(map(int, input("Enter an array:").split()))

count = 1
maximum = 1

for i in range(1, len(arr)):
  if arr[i] > arr[i - 1]:
    count += 1
  else:
    count = 1

  maximum = max(maximum, count)

print("Longest Increasing Contiguous Subarray:",maximum)
```

```python
#Find the Longest Decreasing Contiguous Subarray

#Input
#9 8 7 5 6 4 3 2
#Output
#4

arr = list(map(int, input("Enter an array:").split()))

count = 1
maximum = 1

for i in range(1, len(arr)):
  if arr[i] < arr[i - 1]:
    count += 1
  else:
    count = 1

  maximum = max(maximum, count)

print("Longest Decreasing Contiguous Subarray:",maximum)
```

```python
#Count Number of Peaks

arr = list(map(int, input("Enter an array:").split()))
count = 0

for i in range(1, len(arr)-1):
  if arr[i] > arr[i-1] and arr[i] > arr[i+1]:
    count += 1

print("Number of Peaks:",count)
```

```python
#Longest Zig-Zag Subarray

#Example

#1 3 2 5 4 6

#Output

#5

arr = list(map(int, input("Enter array elements: ").split()))

count = 1
maximum = 1

for i in range(1, len(arr) - 1):

    if (arr[i] > arr[i - 1] and arr[i] > arr[i + 1]) or \
       (arr[i] < arr[i - 1] and arr[i] < arr[i + 1]):

        count += 1

    else:
        count = 2

    maximum = max(maximum, count)

print("Longest Zig-Zag:", maximum)
#
```

```python
#Longest Bitonic Subarray
#(Bitonic = Increasing then Decreasing)

#Input

#12 4 78 90 45 23

#Output

#5

arr = list(map(int, input("Enter array elements: ").split()))

n = len(arr)

inc = [1] * n
dec = [1] * n

for i in range(1, n):

    if arr[i] > arr[i - 1]:
        inc[i] = inc[i - 1] + 1

for i in range(n - 2, -1, -1):

    if arr[i] > arr[i + 1]:
        dec[i] = dec[i + 1] + 1

maximum = 1

for i in range(n):

    maximum = max(maximum, inc[i] + dec[i] - 1)

print("Longest Bitonic Length:", maximum)
```
