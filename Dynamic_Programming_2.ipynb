```python
#Maximum Sum of Non-Adjacent Elements
#Problem: Given an array of integers, find the maximum sum of elements such that no two selected elements are adjacent.
#This is similar to House Robber, but it is a standard DP formulation often used in interviews.
#Example input
#arr = [3, 2, 7, 10]


#Expected output
#13

def max_sum(arr):
  n = len(arr)

  if n == 0:
    return 0

  dp = [0] * n
  dp[0] = arr[0]
  dp[1] = max(arr[0],arr[1])

  for i in range(2,n):
    take = arr[i] + dp[i - 2]
    not_take = dp[i - 1]
    dp[i] = max(take,not_take)

  return dp[n - 1]

arr = [3, 2, 7, 10]
print(max_sum(arr))
```

```python
#Count Ways to Reach the N-th Stair
#Problem: You can climb either 1, 2, or 3 stairs at a time. Find the number of distinct ways to reach stair n.
#Example input
#n = 4


#Expected output
#7

def climb_stairs(n):
  if n < 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(1,n+1):
    dp[i] = dp[i - 1]

    if i >= 2:
      dp[i] += dp[i - 2]

    if i >= 3:
      dp[i] += dp[i - 3]

  return dp[n]

n = 4
print(climb_stairs(n))
```

```python
#Tribonacci Number
#Problem: Given an integer n, calculate the \(n\)-th Tribonacci number. The sequence starts with 0, 1, 1, and every subsequent number is the sum of the previous three numbers.
#Input
#n = 5


#Expected output
#7


def tribonacci(n):
    if n == 0:
        return 0
    if n <= 2:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 0
    dp[1] = 1
    dp[2] = 1

    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2] + dp[i - 3]

    return dp[n]


n = 5
print(tribonacci(n))
```

```python
#Count Ways to Reach the N-th Step Using 1 or 2 Steps
#Problem: You can climb either 1 or 2 steps at a time, but this time calculate the number of ways using a DP array and explicitly handle the starting position n = 0.
#Input
#n = 5


#Expected output
#8

def stair_ways(n):
  if n < 0:
    return 0

  dp = [0] * (n + 1)
  dp[0] = 1

  for i in range(1,n+1):
    dp[i] += dp[i - 1]

    if i >= 2:
      dp[i] += dp[i - 2]

  return dp[n]

n = 5
print(stair_ways(n))
```

```python
#Print Numbers from 1 to N Using DP
#Problem: Given an integer n, create a DP list that stores numbers from 1 to n and print the list.
#Input
#n = 5


#Expected output
#[1, 2, 3, 4, 5]

def print_num(n):
  if n == 0:
    return 0

  dp = [0] * n

  for i in range(n):
    dp[i] = i + 1

  return dp

n = 5
print(print_num(n))
```
