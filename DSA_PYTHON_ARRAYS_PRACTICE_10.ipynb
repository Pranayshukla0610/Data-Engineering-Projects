```python
#Find the Sum of All Elements

arr = list(map(int, input("Enter an array:").split()))

total = 0

for num in arr:
  total += num

print("Sum of All Elements:",total)
```

```python
#Find the Average of Array Elements

arr = list(map(int, input("Enter an array:").split()))

total = 0

for num in arr:
  total += num

print("Average of Array Elements:",total/len(arr))
```

```python
#Count Even and Odd Numbers

arr = list(map(int, input("Enter an array:").split()))

even = 0
odd = 0

for num in arr:
  if num % 2 == 0:
    even += 1
  else:
    odd += 1

print("Even Numbers:",even)
print("Odd Numbers:",odd)
```

```python
#Count Positive, Negative and Zero Elements

arr = list(map(int, input("Enter an array:").split()))

positive = 0
negative = 0
zero = 0

for num in arr:
  if num > 0:
    positive += 1
  elif num < 0:
    negative -= 1
  else:
    zero += 1

print("Positive Numbers:",positive)
print("Negative Numbers:",negative)
print("Zero Numbers:",zero)
```

```python
#Search an Element (Linear Search)

arr = list(map(int, input("Enter an array:").split()))
target = int(input("Enter a target:"))

found = False

for i in range(len(arr)):
  if arr[i] == target:
    found = True
    print("Element Found at Index:",i)
    break

if not found:
  print("Element Not Found")

```

```python
#Count Frequency of a Given Element

arr = list(map(int, input("Enter an array:").split()))
target = int(input("Enter a target:"))

count = 0

for num in arr:
  if num == target:
    count += 1

print("Frequency of target:",count)
```

```python
#Find First and Last Occurrence

arr = list(map(int, input("Enter an array:").split()))
target = int(input("Enter a target:"))

first = -1
last = -1

for i in range(len(arr)):
  if arr[i] == target:
    if first == -1:
      first = i
    last = i

print("First Occurence:",first)
print("Last Occurence:",last)
```

```python
#Copy One Array into Another

arr = list(map(int, input("Enter an array").split()))

copy_arr = []

for num in arr:
  copy_arr.append(num)

print("Original Array:",arr)
print("Copied Array:",copy_arr)
```
