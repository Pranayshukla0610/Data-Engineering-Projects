```python
#Given an integer array, return the running sum of the array.

#Input

#1 2 3 4

#Output

#1 3 6 10

arr = list(map(int, input("Enter an array:").split()))

running_sum = []
current_sum = 0

for num in arr:
  current_sum += num
  running_sum.append(current_sum)

print("Running Sum of an Array:",running_sum)
```

```python
#Construct the Prefix Sum array.
#Sum from index 0 to i
#Input

#5 1 8 2

#Output

#5 6 14 16

arr = list(map(int, input("Enter an array:").split()))

prefix_sum = []
current_sum = 0

for i in range(len(arr)):
  current_sum += arr[i]
  prefix_sum.append(current_sum)

print("Prefix Sum Array:",prefix_sum)
```

```python
#Construct the Prefix Sum array.
#Sum from index 0 to i
#Input

#5 1 8 2

#Output

#5 6 14 16

arr = list(map(int, input("Enter an array:").split()))
n = len(arr)

prefix = [0]*n
prefix[0] = arr[0]

for i in range(1,n):
  prefix[i] = prefix[i-1] + arr[i]

print("Prefix Sum Array:",prefix)
```

```python
#Cumulative Sum
#Input

#10 20 30 40

#Output

#10 30 60 100

arr = list(map(int, input("Enter an array:").split()))

cumulative = []
total = 0

for num in arr:
  total += num
  cumulative.append(total)

print("Cumulative Numbers:",cumulative)
```

```python
#Given an array where each element represents the height of a building, calculate how much rainwater can be trapped after raining.
#Input

#0 1 0 2 1 0 1 3 2 1 2 1

#Output

#6

arr = list(map(int, input("Enter heights:").split()))

n = len(arr)
left = [0]*n
right = [0]*n

left[0] = arr[0]

for i in range(1,n):
  left[i] = max(left[i-1],arr[i])

for i in range(n-2, -1, -1):
  right[i] = max(right[i+1], arr[i])

water = 0

for i in range(n):
  water += min(left[i], right[i]) - arr[i]

print("Total Water Trapped:",water)
```

```python
#Pivot Index
#Find the index where

#Left Sum == Right Sum

#Return the first such index.

#If none exists return -1.

#Input

#1 7 3 6 5 6

#Output

#3

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)

left_sum = 0

pivot = -1

total = sum(arr)

for i in range(n):
  right_sum = total-left_sum-arr[i]

  if left_sum == right_sum:
    pivot = i
    break

  left_sum += arr[i]

print("Pivot Indes:",pivot)
```

```python
#Equilibrium Index
#Find an index where

#Sum of left elements = Sum of right elements

#Input

#-7 1 5 2 -4 3 0

#Output

#6

arr = list(map(int, input("Enter an array:").split()))

n = len(arr)
left_sum = 0
equilibrium = -1
total = sum(arr)

for i in range(n):
  right_sum = total - left_sum - arr[i]

  if left_sum == right_sum:
    equilibrium = i
    break

  left_sum += arr[i]

print("Equilibrium Index:",equilibrium)
```
