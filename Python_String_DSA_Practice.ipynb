```python
#Reverse Words in a Sentence

#Input
#I love Python programming
#Output
#programming Python love I

sentence = input("Enter a sentence:")

words = sentence.split()

reversed_words = []

for i in range(len(words)-1,-1,-1):
  reversed_words.append(words[i])

result = " ".join(reversed_words)
print(result)
```

```python
#Find All Duplicate Characters

string = input("Enter a string:")

char_count = {}

for char in string:
  if char in char_count:
    char_count[char] += 1
  else:
    char_count[char] = 1

print("Duplicate Characters:",char_count)
```

```python
string = input("Enter a string:")

frequency = {}

for ch in string:
  frequency[ch] = frequency.get(ch,0) + 1

print("Duplicate Characters")

found = False

for ch in frequency:
  if frequency[ch] > 1:
    print(ch)
    found = True

if not found:
  print("No Duplicate Characters")
```

```python
#Find the Second Most Frequent Character

string = input("Enter a string:")

frequency = {}

for ch in string:
  frequency[ch] = frequency.get(ch,0) + 1

frequencies = list(set(frequency.values()))

if len(frequencies) < 2:
  print("No second most frequent characters")

else:
  frequencies.sort(reverse=True)
  second_most_frequency = frequencies[1]

  found = False
  for ch in frequency:
    if frequency[ch] == second_most_frequency:
      print(ch)
      found = True
      break

  if not found:
    print("No second most frequent characters")
```

```python
#Replace Multiple Spaces with a Single Space

#Input
#Python    is     very   easy
#Output
#Python is very easy

sentence = input("Enter a sentence:")

words = sentence.split()

result = ''

for word in words:
  if result == "":
    result = word
  else:
    result = result + " " + word

print("Updated Sentence:",result)
```

```python
#Count Occurrences of a Substring

string = input("Enter a string:")
substring = input("Enter a substring:")

count = 0

for i in range(len(string) - len(substring) + 1):
  part = string[i:i+len(substring)]

  if part == substring:
    count += 1

print("Occurences:",count)
```
