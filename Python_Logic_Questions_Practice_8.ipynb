```python
#Find the First Non-Repeated Character in a String

text = input("Enter a text:")

frequency = {}

for char in text:
  if char in frequency:
    frequency[char] += 1
  else:
    frequency[char] = 1

for char in text:
  if frequency[char] == 1:
    print("First Non-Repeated Character:",char)
    break
```

```python
#Reverse Each Word in a Sentence

text = input("Enter a sentence:")

words = text.split()
reverse = []

for word in words:
  reverse.append(word[::-1])

print(' '.join(reverse))
```

```python
#Find All Pairs with Given Sum

arr = list(map(int, input("Enter array: ").split()))
target = int(input("Enter target:"))


seen = set()

for i in range(len(arr)):
  complement = target - arr[i]
  if complement in seen:
    print((complement, arr[i]))

  seen.add(arr[i])
```

```python
#Find the Most Frequent Elementa

arr = list(map(int, input("Enter array:").split()))

frequency = {}

for num in arr:
  frequency[num] = frequency.get(num,0)+1

maximum = 0
answer = None

for key, value in frequency.items():
  if value > maximum:
    maximum = value
    answer = key

print("Most Frequent Element:",answer)
```

```python
#Move All Zeros to End

numbers = list(map(int, input("Enter array:").split()))

result = []
zero_count = 0

for num in numbers:
  if num == 0:
    zero_count += 1
  else:
    result.append(num)

result.extend([0] * zero_count)

print(result)
```

```python
#Find All Duplicate Elements

numbers = list(map(int, input("Enter an array:").split()))

seen = set()
duplicate = set()

for num in numbers:
  if num in seen:
    duplicate.add(num)

  else:
    seen.add(num)

print(list(duplicate))
```

```python
#Find Maximum Consecutive Ones

numbers = list(map(int, input("Enter an array:").split()))

count = 0
maximum = 0

for num in numbers:
  if numbers == 1:
    count += 1
    maximum = max(count,maximum)

  else:
    count = 0

print("Maximum Consecutive Ones",maximum)
```
