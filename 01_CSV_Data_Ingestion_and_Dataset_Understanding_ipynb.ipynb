Different files may use different character encodings.

A common error is:

UnicodeDecodeError

```python
encodings = [
    "utf-8",
    "latin1",
    "cp1252"
]

for encoding in encodings:

    try:

        df = pd.read_csv(
            "/content/ecommerce_sales_34500-selected-columns.csv",
            encoding=encoding
        )

        print(
            "Successfully loaded:",
            encoding
        )

        break

    except UnicodeDecodeError:

        print(
            "Failed:",
            encoding
        )
```

LARGE CSV FILES

This is especially important for Data Engineering.

Suppose:

sales.csv

contains:

50 million rows

```python
chunks = pd.read_csv('/content/ecommerce_sales_34500-selected-columns.csv',chunksize=100000)
```

```python
for chunk in chunks:
  print(
      "Chunk Shape:",
      chunk.shape
  )
```

Process Each Chunk

```python
total_price = 0

for chunk in pd.read_csv(
    "/content/ecommerce_sales_34500-selected-columns.csv",
    chunksize=100000
):

    chunk_total = chunk["price"].sum()

    total_price += chunk_total

print(
    "Total Price:",
    total_price
)
```
