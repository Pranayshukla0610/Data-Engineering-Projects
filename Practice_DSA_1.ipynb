```python
#Two Sum
#Difficulty: Intermediate
#Pattern: Hash map
#Problem statement
#Given an array of integers nums and an integer target, return the indices of two elements whose sum equals the target.

#Assume exactly one solution exists, and you cannot use the same element twice.
#Example input
#nums = [2, 7, 11, 15]target = 9



#Expected output
#[0, 1]

def two_sum(nums, target):
  n = len(nums)
  if n == 0:
    return 0

  seen = {}
  for i in range(n):
    needed = target - nums[i]
    if needed in seen:
      return [seen[needed],i]

    seen[nums[i]] = i
  return []

nums = [2, 7, 11, 15]
target = 9
print(two_sum(nums,target))
```

```python
#Maximum Subarray Sum
#Difficulty: Intermediate
#Pattern: Kadane’s algorithm
#Problem statement
#Given an array containing positive and negative integers, find the maximum sum of any non-empty contiguous subarray.
#Example input
#nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]



#Expected output
#6

def max_sum(nums):
  n = len(nums)
  if n == 0:
    return 0

  max_sum = nums[0]
  curr_sum = nums[0]

  for i in range(n):
    curr_sum = max(nums[i], curr_sum + nums[i])
    max_sum = max(max_sum, curr_sum)

  return max_sum

nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(max_sum(nums))
```

```python
#Maximum Subarray Sum
#Difficulty: Intermediate
#Pattern: Kadane’s algorithm
#Problem statement
#Given an array containing positive and negative integers, find the maximum sum of any non-empty contiguous subarray.
#Example input
#nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]



#Expected output
#6

def max_sum(nums):
  n = len(nums)

  if n == 0:
    return 0

  dp = [0] * n
  dp[0] = nums[0]

  for i in range(1,n):
    dp[i] = max(dp[i - 1] + nums[i], nums[i])

  return max(dp)

nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(max_sum(nums))
```

```python
#Sort Colors
#Difficulty: Intermediate
#Pattern: Dutch National Flag algorithm
#Problem statement
#Given an array containing only 0, 1, and 2, sort the array in-place so that all zeroes come first, followed by ones, then twos.

#Do not use the built-in sorting function.
#Example input
#nums = [2, 0, 2, 1, 1, 0]



#Expected output
#[0, 0, 1, 1, 2, 2]

def sort_colors(nums):
    low = 0
    mid = 0
    high = len(nums) - 1

    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1

        elif nums[mid] == 1:
            mid += 1

        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1

    return nums


nums = [2, 0, 2, 1, 1, 0]

print(sort_colors(nums))
```

```python
#Best Time to Buy and Sell Stock
#Difficulty: Intermediate
#Pattern: Greedy, single-pass array traversal
#Problem statement
#Given an array where prices[i] is a stock's price on day i, choose one day to buy and a later day to sell.

#Return the maximum possible profit. If no profit is possible, return zero.
#Example input
#prices = [7, 1, 5, 3, 6, 4]



#Expected output
#5

def max_profit(prices):
    min_price = prices[0]
    max_profit = 0

    for price in prices:
        min_price = min(min_price, price)

        profit = price - min_price
        max_profit = max(max_profit, profit)

    return max_profit


prices = [7, 1, 5, 3, 6, 4]

print(max_profit(prices))
```
