```python
#Find the Longest Palindromic Substring

string = input("Enter a String:")

n = len(string)

longest = ""

def expand(left, right):
  while left >= 0 and right < n and string[left] == string[right]:
    left -= 1
    right += 1

  return string[left+1:right]

for i in range(n):
  odd = expand(i,i)
  even = expand(i,i+1)

  if len(odd) > len(longest):
    longest = odd

  if len(even) > len(longest):
    longest = even

print("Longest Palindromic Substring:", longest)
```

```python
#Replace Multiple Spaces with a Single Space

string = input("Enter a String:")

words = string.split()

result = ""

for word in words:
  if result =="":
    result = word
  else:
    result = result + " " + word

print("Updated Sentence:",result)
```

```python
#Check if Two Strings are Isomorphic
#Two strings are isomorphic if characters in the first string can be mapped to characters in the second string while maintaining the same pattern.

string1 = input("Enter first string: ")
string2 = input("Enter second string: ")

if len(string1) != len(string2):
    print("Strings are not Isomorphic")

else:

    mapping1 = {}
    mapping2 = {}

    isomorphic = True

    for i in range(len(string1)):

        ch1 = string1[i]
        ch2 = string2[i]

        # Check mapping from string1 to string2
        if ch1 in mapping1:

            if mapping1[ch1] != ch2:
                isomorphic = False
                break

        else:
            mapping1[ch1] = ch2

        # Check reverse mapping
        if ch2 in mapping2:

            if mapping2[ch2] != ch1:
                isomorphic = False
                break

        else:
            mapping2[ch2] = ch1

    if isomorphic:
        print("Strings are Isomorphic")
    else:
        print("Strings are not Isomorphic")
```

```python
#Longest Substring Without Repeating Characters

string = input("Enter a string: ")

left = 0
max_length = 0
longest = ""

seen = set()

for right in range(len(string)):

    # If character is already present,
    # move left until duplicate is removed
    while string[right] in seen:
        seen.remove(string[left])
        left += 1

    # Add current character
    seen.add(string[right])

    # Calculate window length
    current_length = right - left + 1

    # Update maximum
    if current_length > max_length:
        max_length = current_length
        longest = string[left:right + 1]


print("Longest Substring:", longest)
print("Length:", max_length)
```

```python
#Maximum Number of Vowels in a Substring of Length K

string = input("Enter a string: ")
k = int(input("Enter k: "))

vowels = 'aeiou'

count = 0

for i in range(k):
  if string[i] in vowels:
    count += 1

    max_count = count
    best_start = 0

    for right in range(k,len(string)):
      if string[right].lower() in vowels:
        count += 1

      left = right - k

      if string[left].lower() in vowels:
        count -= 1

      if count > max_count:
        max_count = count
        best_start = left + 1

    best_substring = string[
    best_start:best_start + k
]

print("Maximum Vowels:",max_count)
print("Substring:", best_substring)
```
