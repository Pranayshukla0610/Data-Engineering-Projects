```python
#Sum of an Array Using DP
#Easy · Basic DP
#Problem statement: Given an array of integers, calculate the sum of all elements using a DP array that stores the cumulative sum up to each position.

#Input
#arr = [2, 4, 6, 8]


#Output
#20

def sum_array(arr):
  n = len(arr)
  if n == 0:
    return 0

  dp = [0] * n
  dp[0] = arr[0]

  for i in range(1,n):
    dp[i] = dp[i - 1] + arr[i]

  return dp[n-1]

arr = [2,4,6,8]
print(sum_array(arr))
```

```python
#Minimum Cost to Reach the End of an Array
#Easy · 1D DP
#Problem statement: You are given an array of non-negative costs. Start at index 0. At each move, you can jump one or two positions forward. Find the minimum total cost required to reach the last index.

#The cost of the starting and ending positions is included.

#Input
#cost = [1, 3, 2, 4]


#Output
#7

def min_cost(cost):
  n = len(cost)
  if n <= 0:
    return 0

  dp = [0] * (n)
  dp[0] = cost[0]

  if n == 1:
    return dp[0]

  dp[1] = cost[1] + dp[0]

  for i in range(2, n):
    dp[i] = min(dp[i-1],dp[i-2]) + cost[i]

  return dp[n-1]

cost = [1,3,2,4]
print(min_cost(cost))
```

```python
#Maximum Sum of Non-Adjacent Elements
#Easy · 1D DP
#Problem statement: Given an array of non-negative integers, find the maximum sum you can obtain without selecting two adjacent elements.

#Input
#arr = [2, 7, 9, 3, 1]


#Output
#12


def max_non_adjacent_sum(arr):
    n = len(arr)

    if n == 0:
        return 0

    if n == 1:
        return arr[0]

    dp = [0] * n
    dp[0] = arr[0]
    dp[1] = max(arr[0], arr[1])

    for i in range(2, n):
        take = arr[i] + dp[i - 2]
        skip = dp[i - 1]

        dp[i] = max(take, skip)

    return dp[n - 1]


arr = [2, 7, 9, 3, 1]
print(max_non_adjacent_sum(arr))
```

```python
#Count Ways to Reach the N-th Step
#Easy · Counting DP
#Problem statement: You start at step 0. You can move either one or two steps at a time. Count how many ways you can reach step n.

#This is similar to Climbing Stairs, but let's implement it using a different base-case convention.

#Input
#n = 5


#Output
#8


def count_ways(n):
    dp = [0] * (n + 1)

    dp[0] = 1

    for i in range(1, n + 1):
        dp[i] += dp[i - 1]

        if i >= 2:
            dp[i] += dp[i - 2]

    return dp[n]


n = 5
print(count_ways(n))
```
